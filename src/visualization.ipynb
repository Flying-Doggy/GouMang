```python
import goumang_plot
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import matplotlib.patches as mpatches
```

```python
chromosome_df = pd.read_csv('/home/gujianhui/BLOCKOME/GouMang/test/lt_2/lt_2_mapping_res_chromosomes_annot.tsv',   sep='\t')
chromosome_df['chromosome_id'] = chromosome_df['Unnamed: 0']
```

```python
col_idxes = [i for i in chromosome_df.columns if i.isdigit()]
tmp_df = chromosome_df.sort_values(by='clust', ascending=False)
goumang_plot.heatmap_plot( df=tmp_df , heatmap_cols=col_idxes, linewidths = 0.5 , clust_col='clust', tick_col='chromosome_id' )

# plt.savefig('../img/test_chromosome_comp.svg', format='svg', bbox_inches='tight')
```

```python
goumang_plot.umap_plot( df=chromosome_df  )
# plt.savefig('../img/test_chromosomes_umap.svg', format='svg', bbox_inches='tight')
```
